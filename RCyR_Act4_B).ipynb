```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# SISTEMA EXPERTO SIMPLE PARA MEDIR ESTRÉS

# 1. BASE DE CONOCIMIENTOS

base_conocimientos = [

    # Reglas originales de estrés
    {
        "nombre": "Estrés Alto",
        "condiciones": {
            "horas_sueno": "bajo",
            "ansiedad": "alta",
            "fatiga": "alta"
        },
        "conclusion": "El paciente presenta estrés alto."
    },
    {
        "nombre": "Estrés Medio",
        "condiciones": {
            "horas_sueno": "medio",
            "ansiedad": "media"
        },
        "conclusion": "El paciente presenta estrés medio."
    },
    {
        "nombre": "Estrés Bajo",
        "condiciones": {
            "horas_sueno": "alto",
            "ansiedad": "baja",
            "fatiga": "baja"
        },
        "conclusion": "El paciente presenta estrés bajo."
    },

    # Reglas para el nivel de sueño
    {
        "nombre": "Sueño Bajo",
        "condiciones": {"horas_sueno": "bajo"},
        "conclusion": "El paciente presenta un nivel de sueño bajo."
    },
    {
        "nombre": "Sueño Medio",
        "condiciones": {"horas_sueno": "medio"},
        "conclusion": "El paciente presenta un nivel de sueño medio."
    },
    {
        "nombre": "Sueño Alto",
        "condiciones": {"horas_sueno": "alto"},
        "conclusion": "El paciente presenta un nivel de sueño alto."
    },

    # Reglas para el nivel de ansiedad
    {
        "nombre": "Ansiedad Alta",
        "condiciones": {"ansiedad": "alta"},
        "conclusion": "El paciente presenta ansiedad alta."
    },
    {
        "nombre": "Ansiedad Media",
        "condiciones": {"ansiedad": "media"},
        "conclusion": "El paciente presenta ansiedad media."
    },
    {
        "nombre": "Ansiedad Baja",
        "condiciones": {"ansiedad": "baja"},
        "conclusion": "El paciente presenta ansiedad baja."
    },

    # Reglas para el nivel de fatiga
    {
        "nombre": "Fatiga Alta",
        "condiciones": {"fatiga": "alta"},
        "conclusion": "El paciente presenta fatiga alta."
    },
    {
        "nombre": "Fatiga Media",
        "condiciones": {"fatiga": "media"},
        "conclusion": "El paciente presenta fatiga media."
    },
    {
        "nombre": "Fatiga Baja",
        "condiciones": {"fatiga": "baja"},
        "conclusion": "El paciente presenta fatiga baja."
    }
]

# 2. MEMORIA DE HECHOS (MEMORIA TEMPORAL)

memoria_hechos = {}

# 3. INTERFAZ DE USUARIO

print("===== SISTEMA EXPERTO DE ESTRÉS =====")

memoria_hechos["horas_sueno"] = input(
    "Nivel de sueño (alto/medio/bajo): "
).lower().strip()

memoria_hechos["ansiedad"] = input(
    "Nivel de ansiedad (alta/media/baja): "
).lower().strip()

memoria_hechos["fatiga"] = input(
    "Nivel de fatiga (alta/media/baja): "
).lower().strip()

# 4. MOTOR DE INFERENCIA

def motor_inferencia(hechos, reglas):

    resultados = []

    for regla in reglas:
        condiciones = regla["condiciones"]

        cumple = True

        # Verificar si todas las condiciones se cumplen
        for clave, valor in condiciones.items():

            if hechos.get(clave) != valor:
                cumple = False
                break

        # Guardar las reglas que coinciden
        if cumple:
            resultados.append(regla)

    return resultados

# 5. MÓDULO DE EXPLICACIÓN

def explicar(hechos, resultados):

    print("\n--- EXPLICACIÓN ---")
    print("Hechos analizados:")

    for clave, valor in hechos.items():
        print(f"- {clave}: {valor}")

    print("\nReglas aplicadas:")

    for regla in resultados:
        print(f"- {regla['nombre']}")

# EJECUCIÓN DEL SISTEMA

resultado = motor_inferencia(memoria_hechos, base_conocimientos)

print("\n===== RESULTADO =====")

# Mostrar el diagnóstico de estrés por separado
reglas_estres = [
    regla for regla in resultado
    if regla["nombre"].startswith("Estrés")
]

if reglas_estres:
    for regla in reglas_estres:
        print(regla["conclusion"])
else:
    print("No se pudo determinar el nivel de estrés con las reglas disponibles.")

# Mostrar los resultados de sueño, ansiedad y fatiga
print("\n===== EVALUACIÓN DE FACTORES =====")

for regla in resultado:
    if not regla["nombre"].startswith("Estrés"):
        print(regla["conclusion"])

explicar(memoria_hechos, resultado)
```
