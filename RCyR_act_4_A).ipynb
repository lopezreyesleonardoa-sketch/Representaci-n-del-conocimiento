```python

#Leonardo Alfredo López Reyes 2181755
import math

# Regresión logística
x1 = 2
x2 = 90
x3 = 3
x4 = 85
x5 = 10

# X
X = -4.5 - 2.345*x1 + 0.23198*x2 - 0.983345*x3 - 0.00009812*x4 + 0.00004231*x5

# Probabilidad
y = 1 / (1 + math.exp(-X))
Porcentaje=y*100
print("Valor de X:", X)
print("Probabilidad:", y)
print("Porcentaje:", Porcentaje, "%")
```
